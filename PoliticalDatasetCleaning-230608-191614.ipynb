```python
def rename_columns(df, old_names, new_names):
    """
    Renames columns in a DataFrame.
    
    Args:
        df (pd.DataFrame): The DataFrame to rename columns in.
        old_names (list): List of old column names.
        new_names (list): List of new column names.
        
    Returns:
        pd.DataFrame: DataFrame with renamed columns.
    """
    
    if len(old_names) != len(new_names):
        raise ValueError("Number of old names and new names should be the same.")
    
    mapping = {old: new for old, new in zip(old_names, new_names)}
    df.rename(columns=mapping, inplace=True)
    
    return df
```

```python
import re

def add_comma_to_options(column):
    pattern = r"(TV News|Social Media|Print Newspapers|Online News Websites|Radio|Other \(Please specify\))"
    matches = re.findall(pattern, str(column))
    result = ", ".join(matches)
    return result
```

```python
df['political_information'] = df['political_information'].astype(str).apply(add_comma_to_options)
```
